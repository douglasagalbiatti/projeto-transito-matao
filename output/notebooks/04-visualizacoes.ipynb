```python
# 6.1: Mapa Interativo de Sinistros em Matão/SP (com estilo Carto Positron livre de bloqueio)
import pandas as pd
import plotly.express as px
import os
import sys

sys.path.append('..')
from src.metricas import ORDEM_GRAVIDADE, SEM_FERIDOS, SEM_INFORMACAO

# Carregar dados limpos
df = pd.read_csv('../data/matao_limpo.csv', parse_dates=['data'])

# Filtrar coordenadas válidas na região de Matão; notificações repetem poucas coordenadas e ficam fora
df_mapa = df[df['tipo_registro'] != 'NOTIFICACAO'].dropna(subset=['latitude', 'longitude']).copy()
filtro_matao = (
    (df_mapa['latitude'].between(-21.80, -21.40)) & 
    (df_mapa['longitude'].between(-48.60, -48.10))
)
df_mapa = df_mapa[filtro_matao].copy()
df_mapa['data_texto'] = df_mapa['data'].dt.strftime('%d/%m/%Y')

# Ordem e paleta de cores
ordem_gravidade = ORDEM_GRAVIDADE
cores_gravidade = {
    'Com óbito': '#d90429',        # Vermelho vivo
    'Com feridos graves': '#f77f00', # Laranja
    'Com feridos leves': '#e5b124',  # Dourado
    SEM_FERIDOS: '#2b5c8f',          # Azul
    SEM_INFORMACAO: '#adb5bd'        # Cinza
}

# Construir o Mapa usando 'carto-positron'
fig = px.scatter_map(
    df_mapa,
    lat='latitude',
    lon='longitude',
    color='gravidade',
    category_orders={'gravidade': ordem_gravidade},
    color_discrete_map=cores_gravidade,
    hover_name='tipo_sinistro',
    hover_data={
        'latitude': False,
        'longitude': False,
        'gravidade': True,
        'data_texto': True,
        'hora': True,
        'tipo_via': True,
        'logradouro': True,
        'total_vitimas': True
    },
    zoom=12.2,
    center={"lat": -21.6035, "lon": -48.3645},
    map_style='carto-positron',  # << ESTILO PROFISSIONAL E SEM BLOQUEIO 403
    height=680,
    title=f"Mapeamento dos Sinistros de Trânsito em Matão/SP (dados até {df['data'].max():%d/%m/%Y})"
)

fig.update_layout(
    margin=dict(l=0, r=0, t=40, b=0),
    legend=dict(
        title=dict(text='<b>Gravidade</b>'),
        bgcolor='rgba(255, 255, 255, 0.9)',
        bordercolor='#cccccc',
        borderwidth=1,
        x=0.01,
        y=0.99
    )
)

# Salvar como página HTML
caminho_html = '../output/graficos/mapa_acidentes.html'
fig.write_html(caminho_html)

print("--- Mapa Atualizado com Sucesso ---")
print(f"Total de sinistros mapeados: {len(df_mapa)}")
print(f"Arquivo interativo salvo em: {caminho_html}")

fig.show()

# Versão estática do mapa, sem notificações e sem mapa de fundo
import matplotlib.pyplot as plt

fig_estatica, ax = plt.subplots(figsize=(10, 7.5), dpi=200)
for nivel in reversed(ordem_gravidade):
    pontos = df_mapa[df_mapa['gravidade'] == nivel]
    fatal = nivel == 'Com óbito'
    ax.scatter(
        pontos['longitude'], pontos['latitude'],
        s=45 if fatal else 16,
        c=cores_gravidade[nivel],
        edgecolors='black' if fatal else 'none',
        linewidths=0.5,
        alpha=1 if fatal else 0.7,
        label=f'{nivel} ({len(pontos):,})'.replace(',', '.'),
    )
ax.set_title(f"Sinistros de Trânsito em Matão/SP (dados até {df['data'].max():%d/%m/%Y})", fontweight='bold')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.ticklabel_format(useOffset=False)
ax.xaxis.set_major_formatter(lambda valor, _: f'{valor:.2f}'.replace('.', ','))
ax.yaxis.set_major_formatter(lambda valor, _: f'{valor:.3f}'.replace('.', ','))
ax.grid(linestyle=':', alpha=0.5)
ax.legend(title='Gravidade', loc='upper center', bbox_to_anchor=(0.5, -0.09), ncol=3, fontsize=8)
fig_estatica.tight_layout()
fig_estatica.savefig('../output/graficos/mapa_estatico_matao.png', bbox_inches='tight')
plt.show()
```

```python
# 6.2: Ranking das Vias Mais Críticas e Exportação do Relatório Anual (Seção 8)

# 1. Ranking das 10 Vias com Mais Sinistros em Matão
ranking_vias = df[df['tipo_registro'] != 'NOTIFICACAO'].groupby('logradouro').agg(
    total_sinistros=('id_sinistro', 'count'),
    obitos=('mortos', 'sum'),
    feridos_graves=('feridos_graves', 'sum'),
    feridos_leves=('feridos_leves', 'sum')
).sort_values(by='total_sinistros', ascending=False).head(10)

print("--- TOP 10 VIAS COM MAIS SINISTROS EM MATÃO (SEM NOTIFICAÇÕES) ---")
print(ranking_vias)

# 2. Resumo Estatístico Anual Consolidado (Conforme Seção 8 do Guia)
resumo_anual = df.groupby('ano').agg(
    total_sinistros=('id_sinistro', 'count'),
    obitos=('mortos', 'sum'),
    feridos_graves=('feridos_graves', 'sum'),
    feridos_leves=('feridos_leves', 'sum'),
    ilesos=('ilesos', 'sum')
).reset_index()

caminho_relatorio = '../output/relatorios/resumo_anual_matao.csv'
os.makedirs('../output/relatorios', exist_ok=True)
resumo_anual.to_csv(caminho_relatorio, index=False)

print(f"\n--- Relatório Anual Salvo com Sucesso em: {caminho_relatorio} ---")
resumo_anual
```
